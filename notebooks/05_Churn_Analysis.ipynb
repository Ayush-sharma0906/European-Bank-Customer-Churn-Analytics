```python
import pandas as pd

df = pd.read_csv("../data/European_Bank_Processed.csv")

df.head()
```

```python
total_customers = len(df)
churned_customers = df["Exited"].sum()
retained_customers = total_customers - churned_customers

overall_churn_rate = churned_customers / total_customers * 100
retention_rate = retained_customers / total_customers * 100

print("Total Customers :", total_customers)
print("Churned Customers:", churned_customers)
print("Retained Customers:", retained_customers)
print("Overall Churn Rate:", round(overall_churn_rate, 2), "%")
print("Retention Rate    :", round(retention_rate, 2), "%")
```

```python
geography_churn = (
    df.groupby("Geography")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

geography_churn["ChurnRate"] = geography_churn["mean"] * 100

geography_churn["ChurnContribution"] = (
    geography_churn["sum"] / churned_customers * 100
)

geography_churn = geography_churn.drop(columns="mean")

geography_churn
```

```python
age_churn = (
    df.groupby("AgeGroup")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

age_churn["ChurnRate"] = age_churn["mean"] * 100

age_churn["ChurnContribution"] = (
    age_churn["sum"] / churned_customers * 100
)

age_churn = age_churn.drop(columns="mean")

age_churn
```

```python
credit_churn = (
    df.groupby("CreditScoreBand")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

credit_churn["ChurnRate"] = credit_churn["mean"] * 100

credit_churn["ChurnContribution"] = (
    credit_churn["sum"] / churned_customers * 100
)

credit_churn = credit_churn.drop(columns="mean")

credit_churn
```

```python
tenure_churn = (
    df.groupby("TenureGroup")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

tenure_churn["ChurnRate"] = tenure_churn["mean"] * 100

tenure_churn["ChurnContribution"] = (
    tenure_churn["sum"] / churned_customers * 100
)

tenure_churn = tenure_churn.drop(columns="mean")

tenure_churn
```

```python
balance_churn = (
    df.groupby("BalanceSegment")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

balance_churn["ChurnRate"] = balance_churn["mean"] * 100

balance_churn["ChurnContribution"] = (
    balance_churn["sum"] / churned_customers * 100
)

balance_churn = balance_churn.drop(columns="mean")

balance_churn
```

```python
profile_comparison = (
    df.groupby("Exited")[
        ["CreditScore", "Age", "Tenure",
         "Balance", "NumOfProducts", "EstimatedSalary"]
    ]
    .mean()
    .round(2)
)

profile_comparison.index = ["Retained", "Churned"]

profile_comparison
```

```python
engagement_churn = (
    df.groupby("EngagementStatus")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

engagement_churn["ChurnRate"] = engagement_churn["mean"] * 100

engagement_churn["ChurnContribution"] = (
    engagement_churn["sum"] / churned_customers * 100
)

engagement_churn = engagement_churn.drop(columns="mean")

engagement_churn
```

```python
product_churn = (
    df.groupby("NumOfProducts")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

product_churn["ChurnRate"] = product_churn["mean"] * 100

product_churn["ChurnContribution"] = (
    product_churn["sum"] / churned_customers * 100
)

product_churn = product_churn.drop(columns="mean")

product_churn
```

```python
gender_churn = (
    df.groupby("Gender")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

gender_churn["ChurnRate"] = gender_churn["mean"] * 100

gender_churn["ChurnContribution"] = (
    gender_churn["sum"] / churned_customers * 100
)

gender_churn = gender_churn.drop(columns="mean")

gender_churn
```

```python
risk_profile = (
    df.groupby(
        ["Geography", "AgeGroup",
         "EngagementStatus", "BalanceSegment"]
    )["Exited"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

risk_profile["ChurnRate"] = risk_profile["mean"] * 100

risk_profile["ChurnContribution"] = (
    risk_profile["sum"] / churned_customers * 100
)

risk_profile = risk_profile.drop(columns="mean")

risk_profile = risk_profile.sort_values(
    "ChurnRate",
    ascending=False
)

risk_profile.head(15)
```

```python
reliable_risk_profiles = (
    risk_profile[
        risk_profile["count"] >= 50
    ]
    .sort_values(
        "ChurnRate",
        ascending=False
    )
)

reliable_risk_profiles.head(15)
```

```python
geo_age_churn = (
    df.groupby(["Geography", "AgeGroup"])["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

geo_age_churn["ChurnRate"] = geo_age_churn["mean"] * 100

geo_age_churn["ChurnContribution"] = (
    geo_age_churn["sum"] / churned_customers * 100
)

geo_age_churn = geo_age_churn.drop(columns="mean")

geo_age_churn = geo_age_churn.sort_values(
    "ChurnRate",
    ascending=False
)

geo_age_churn
```

```python
# 1. Overall Churn Rate
overall_churn_rate = df["Exited"].mean() * 100

# 2. Segment Churn Rate
segment_churn_rate = {
    "Germany": geography_churn.loc[
        geography_churn["Geography"] == "Germany", "ChurnRate"
    ].iloc[0],

    "Age 46-60": age_churn.loc[
        age_churn["AgeGroup"] == "46-60", "ChurnRate"
    ].iloc[0],

    "High-balance": balance_churn.loc[
        balance_churn["BalanceSegment"] == "High-balance", "ChurnRate"
    ].iloc[0]
}

# 3. High-Value Churn Ratio
high_value_customers = (
    df["BalanceSegment"] == "High-balance"
).sum()

high_value_churners = (
    (df["BalanceSegment"] == "High-balance") &
    (df["Exited"] == 1)
).sum()

high_value_churn_ratio = (
    high_value_churners / high_value_customers * 100
)

# 4. Geographic Risk Index
# Germany churn rate relative to overall churn rate
germany_churn_rate = segment_churn_rate["Germany"]

geographic_risk_index = (
    germany_churn_rate / overall_churn_rate
)

# 5. Engagement Drop Indicator
active_churn_rate = engagement_churn.loc[
    engagement_churn["EngagementStatus"] == "Active",
    "ChurnRate"
].iloc[0]

inactive_churn_rate = engagement_churn.loc[
    engagement_churn["EngagementStatus"] == "Inactive",
    "ChurnRate"
].iloc[0]

engagement_drop_indicator = (
    inactive_churn_rate - active_churn_rate
)

kpi_summary = pd.DataFrame({
    "KPI": [
        "Overall Churn Rate",
        "Germany Churn Rate",
        "Age 46-60 Churn Rate",
        "High-balance Churn Rate",
        "High-Value Churn Ratio",
        "Geographic Risk Index",
        "Engagement Drop Indicator"
    ],
    "Value": [
        overall_churn_rate,
        germany_churn_rate,
        segment_churn_rate["Age 46-60"],
        segment_churn_rate["High-balance"],
        high_value_churn_ratio,
        geographic_risk_index,
        engagement_drop_indicator
    ]
})

kpi_summary["Value"] = kpi_summary["Value"].round(2)

kpi_summary
```

```python
# Save Churn Analysis tables

geography_churn.to_csv(
    "../outputs/tables/geography_churn_summary.csv",
    index=False
)

age_churn.to_csv(
    "../outputs/tables/age_churn_summary.csv",
    index=False
)

credit_churn.to_csv(
    "../outputs/tables/credit_churn_summary.csv",
    index=False
)

tenure_churn.to_csv(
    "../outputs/tables/tenure_churn_summary.csv",
    index=False
)

balance_churn.to_csv(
    "../outputs/tables/balance_churn_summary.csv",
    index=False
)

engagement_churn.to_csv(
    "../outputs/tables/engagement_churn_summary.csv",
    index=False
)

product_churn.to_csv(
    "../outputs/tables/product_churn_summary.csv",
    index=False
)

gender_churn.to_csv(
    "../outputs/tables/gender_churn_summary.csv",
    index=False
)

geo_age_churn.to_csv(
    "../outputs/tables/geo_age_churn_summary.csv",
    index=False
)

profile_comparison.to_csv(
    "../outputs/tables/profile_comparison.csv"
)

reliable_risk_profiles.to_csv(
    "../outputs/tables/reliable_risk_profiles.csv",
    index=False
)

kpi_summary.to_csv(
    "../outputs/tables/kpi_summary.csv",
    index=False
)

print("✅ All Churn Analysis tables saved successfully!")
```
