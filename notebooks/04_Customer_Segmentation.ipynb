```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/European_Bank_Processed.csv")

print("Shape:", df.shape)
df.head()
```

```python
geography_summary = (
    df.groupby("Geography")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

geography_summary["ChurnRate"] = geography_summary["mean"] * 100

geography_summary = geography_summary.drop(columns="mean")

geography_summary
```

```python
age_summary = (
    df.groupby("AgeGroup")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

age_summary["ChurnRate"] = age_summary["mean"] * 100

age_summary = age_summary.drop(columns="mean")

age_summary
```

```python
age_order = ["<30", "30-45", "46-60", "60+"]

age_summary["AgeGroup"] = pd.Categorical(
    age_summary["AgeGroup"],
    categories=age_order,
    ordered=True
)

age_summary = age_summary.sort_values("AgeGroup")

age_summary
```

```python
credit_summary = (
    df.groupby("CreditScoreBand")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

credit_summary["ChurnRate"] = credit_summary["mean"] * 100

credit_summary = credit_summary.drop(columns="mean")

credit_order = ["Low", "Medium", "High"]

credit_summary["CreditScoreBand"] = pd.Categorical(
    credit_summary["CreditScoreBand"],
    categories=credit_order,
    ordered=True
)

credit_summary = credit_summary.sort_values("CreditScoreBand")

credit_summary
```

```python
tenure_summary = (
    df.groupby("TenureGroup")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

tenure_summary["ChurnRate"] = tenure_summary["mean"] * 100

tenure_summary = tenure_summary.drop(columns="mean")

tenure_order = ["New", "Mid-term", "Long-term"]

tenure_summary["TenureGroup"] = pd.Categorical(
    tenure_summary["TenureGroup"],
    categories=tenure_order,
    ordered=True
)

tenure_summary = tenure_summary.sort_values("TenureGroup")

tenure_summary
```

```python
balance_summary = (
    df.groupby("BalanceSegment")["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

balance_summary["ChurnRate"] = balance_summary["mean"] * 100

balance_summary = balance_summary.drop(columns="mean")

balance_order = [
    "Zero-balance",
    "Low-balance",
    "High-balance"
]

balance_summary["BalanceSegment"] = pd.Categorical(
    balance_summary["BalanceSegment"],
    categories=balance_order,
    ordered=True
)

balance_summary = balance_summary.sort_values("BalanceSegment")

balance_summary
```

```python
geography_summary["ChurnContribution"] = (
    geography_summary["sum"] / df["Exited"].sum() * 100
)

geography_summary
```

```python
age_summary["ChurnContribution"] = (
    age_summary["sum"] / df["Exited"].sum() * 100
)

age_summary
```

```python
credit_summary["ChurnContribution"] = (
    credit_summary["sum"] / df["Exited"].sum() * 100
)

credit_summary
```

```python
tenure_summary["ChurnContribution"] = (
    tenure_summary["sum"] / df["Exited"].sum() * 100
)

tenure_summary
```

```python
balance_summary["ChurnContribution"] = (
    balance_summary["sum"] / df["Exited"].sum() * 100
)

balance_summary
```

```python
segment_priority = pd.DataFrame({
    "SegmentType": [
        "Geography",
        "Age",
        "Credit Score",
        "Tenure",
        "Balance"
    ],
    "HighestRiskSegment": [
        "Germany",
        "46-60",
        "Low",
        "New",
        "High-balance"
    ],
    "HighestChurnRate": [
        geography_summary["ChurnRate"].max(),
        age_summary["ChurnRate"].max(),
        credit_summary["ChurnRate"].max(),
        tenure_summary["ChurnRate"].max(),
        balance_summary["ChurnRate"].max()
    ]
})

segment_priority
```

```python
segment_priority.to_csv(
    "../outputs/tables/segment_priority_summary.csv",
    index=False
)

segment_priority
```
