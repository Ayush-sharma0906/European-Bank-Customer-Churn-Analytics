```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/European_Bank.csv")
print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df.dtypes
```

```python
df.nunique()
```

```python
df_clean = df.drop(
    columns=["Year", "CustomerId", "Surname"]
).copy()

df_clean.head()
```

```python
print("Original shape:", df.shape)
print("Clean shape:", df_clean.shape)
```

```python
def age_group(age):
    if age < 30:
        return "<30"
    elif age <= 45:
        return "30-45"
    elif age <= 60:
        return "46-60"
    else:
        return "60+"

df_clean["AgeGroup"] = df_clean["Age"].apply(age_group)

df_clean["AgeGroup"].value_counts().sort_index()
```

```python
df_clean["CreditScoreBand"] = pd.cut(
    df_clean["CreditScore"],
    bins=[-np.inf, 584, 718, np.inf],
    labels=["Low", "Medium", "High"]
)
df_clean["CreditScoreBand"].value_counts().sort_index()
```

```python
df_clean["CreditScoreBand"].isna().sum()
```

```python
def tenure_group(tenure):
    if tenure <= 3:
        return "New"
    elif tenure <= 7:
        return "Mid-term"
    else:
        return "Long-term"

df_clean["TenureGroup"] = df_clean["Tenure"].apply(tenure_group)
df_clean["TenureGroup"].value_counts()
```

```python
df_clean["TenureGroup"].isna().sum()
```

```python
def balance_segment(balance):
    if balance == 0:
        return "Zero-balance"
    elif balance <= 97198.54:
        return "Low-balance"
    else:
        return "High-balance"

df_clean["BalanceSegment"] = df_clean["Balance"].apply(balance_segment)
df_clean["BalanceSegment"].value_counts()
```

```python
df_clean["BalanceSegment"].isna().sum()
```

```python
df_clean["EngagementStatus"] = df_clean["IsActiveMember"].map({
    0: "Inactive",
    1: "Active"
})
df_clean["EngagementStatus"].value_counts()
```

```python
df_clean["EngagementStatus"].isna().sum()
```

```python
print("Shape:", df_clean.shape)

print("\nColumns:")
print(df_clean.columns.tolist())

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:", df_clean.duplicated().sum())
```

```python
print(df_clean["Exited"].value_counts())
print("\nChurn rate:", df_clean["Exited"].mean() * 100)
```

```python
df_clean.to_csv("../data/European_Bank_Processed.csv", index=False)

print("Processed dataset saved successfully.")
print("Shape:", df_clean.shape)
```
